Q3

```python
x=np.array([1,2,3,4,5,1,2,1,1,1])
```

```python
def findind(arr):
  vals,counts=np.unique(arr,return_counts=True)
  countmax=counts.max()
  modes=vals[counts==countmax]
  for  mode in modes:
    indices=np.where(arr==mode)[0]
    return indices
```

```python
findind(x)
```

```python
y=np.array([1,1,1,2,3,4,2,4,3,3])
```

```python
findind(y)
```
